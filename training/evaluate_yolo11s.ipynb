```python
from ultralytics import YOLO
from pathlib import Path

model_path = Path(
    r"D:\FoodLens_AI\training\runs\runs\foodlens_yolo11s\weights\best.pt"
)

model = YOLO(str(model_path))

print("Model exists:", model_path.exists())
print("Model loaded successfully!")
```

```python
test_yaml = r"D:\FoodLens_AI\data\processed\foodlens_20\data.yaml"

metrics = model.val(
    data=test_yaml,
    split="test",
    imgsz=640,
    batch=4,
    device=0,
    workers=4,
    plots=True
)

print("Test evaluation completed!")
```

```python
print("========== YOLO11s TEST RESULTS ==========")

print(f"mAP50:      {metrics.box.map50:.4f}")
print(f"mAP50-95:   {metrics.box.map:.4f}")
print(f"Precision:  {metrics.box.mp:.4f}")
print(f"Recall:     {metrics.box.mr:.4f}")
```

```python
names = model.names

print("========== PER-CLASS RESULTS ==========")

for i, name in names.items():
    print(
        f"{i:2d}. {name:25s} | "
        f"mAP50: {metrics.box.ap50[i]:.3f} | "
        f"mAP50-95: {metrics.box.ap[i]:.3f}"
    )
```
