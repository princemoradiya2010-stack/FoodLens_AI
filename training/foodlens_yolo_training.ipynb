```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

print("YOLO11n loaded successfully")
```

```python
print(model.model)
```

```python
results = model.train(
    data="../data/processed/foodlens_20/data.yaml",

    # Training
    epochs=50,
    imgsz=640,

    # RTX 4050 6 GB
    batch=8,
    device=0,

    # Performance
    workers=4,
    cache=False,

    # Training control
    patience=10,

    # Save results
    project="../results",
    name="foodlens_yolo11n",

    # Model settings
    pretrained=True,

    # Augmentation
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=10,
    translate=0.1,
    scale=0.5,
    fliplr=0.5,

    # Reproducibility
    seed=42
)
```
