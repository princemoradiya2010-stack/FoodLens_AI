```python
from pathlib import Path
from collections import Counter

dataset_dir = Path(
    r"D:\FoodLens_AI\data\processed\foodlens_20"
)

labels_dir = dataset_dir / "train" / "labels"
data_yaml = dataset_dir / "data.yaml"

# Read class names from data.yaml
import yaml

with open(data_yaml, "r", encoding="utf-8") as f:
    data = yaml.safe_load(f)

class_names = data["names"]

print("Number of classes:", len(class_names))
print()

# Count objects
object_counts = Counter()

# Count images containing each class
image_counts = Counter()

label_files = list(labels_dir.glob("*.txt"))

for label_file in label_files:

    classes_in_image = set()

    with open(label_file, "r") as f:

        for line in f:

            parts = line.strip().split()

            if len(parts) != 5:
                continue

            class_id = int(parts[0])

            object_counts[class_id] += 1
            classes_in_image.add(class_id)

    for class_id in classes_in_image:
        image_counts[class_id] += 1


print("========== TRAINING DATA ==========")

for class_id, class_name in enumerate(class_names):

    print(
        f"{class_id:2d}. "
        f"{class_name:25s} | "
        f"Images: {image_counts[class_id]:4d} | "
        f"Objects: {object_counts[class_id]:4d}"
    )
```

```python
print("========== SORTED BY OBJECT COUNT ==========")

sorted_classes = sorted(
    range(len(class_names)),
    key=lambda x: object_counts[x]
)

for class_id in sorted_classes:

    print(
        f"{class_names[class_id]:25s} | "
        f"{object_counts[class_id]:4d} objects | "
        f"{image_counts[class_id]:4d} images"
    )
```

```python
max_objects = max(object_counts.values())
min_objects = min(object_counts.values())

print("Maximum objects:", max_objects)
print("Minimum objects:", min_objects)
print(
    "Maximum / Minimum ratio:",
    round(max_objects / min_objects, 2)
)
```
