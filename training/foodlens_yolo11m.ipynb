```python
from ultralytics import YOLO

model = YOLO("yolo11m.pt")

print("YOLO11m loaded successfully!")
```

```python
results = model.train(
    data=r"D:\FoodLens_AI\data\processed\foodlens_20\data.yaml",

    epochs=80,
    imgsz=640,
    batch=2,

    device=0,
    workers=2,
    cache=False,

    patience=20,

    project=r"D:\FoodLens_AI\training\runs",
    name="foodlens_yolo11m",

    pretrained=True,

    # Food-friendly augmentation
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,

    degrees=10,
    translate=0.1,
    scale=0.5,
    fliplr=0.5,

    seed=42
)
```
