```python
from ultralytics import YOLO

print("Ultralytics YOLO imported successfully!")
    
```

```python
model = YOLO("yolo11s.pt")

print("YOLO11s loaded successfully!")
```

```python
results = model.train(
    data="../data/processed/foodlens_20/data.yaml",

    epochs=60,
    imgsz=640,

    batch=4,

    device=0,
    workers=4,

    cache=False,

    patience=15,

    project="../runs",
    name="foodlens_yolo11s",

    pretrained=True,

    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,

    degrees=10,
    translate=0.1,
    scale=0.5,
    fliplr=0.5,

    seed=42
)
```
